## Introduction
To ensure safe roads, it is important to know where most accidents happen. However, some roads have more accidents than others simply because they carry more traffic. Research shows that the number of accidents goes up when traffic volume increases, although not always at the same rate (Høye & Hesjevoll, 2020). This means that a list of roads with the most accidents mostly shows the busiest roads, not necessarily the most dangerous ones. To find the roads that are genuinely more dangerous, the number of accidents has to be related to the amount of traffic.

This approach has been used before in the Netherlands. SWOV combined the same datasets we use (BRON and INWEVA) to compare accident risk on national highways between different regions, and looked at whether things like speed limits and the share of trucks could explain the differences (Schepers et al., 2020). In our project we look at individual road sections on the national highways (rijkswegen) in South Holland, managed by Rijkswaterstaat. Knowing which road sections are relatively unsafe helps road authorities decide where safety measures are most needed.

## Objective
To identify national-highway carriageway sections in South Holland that experienced relatively high crash frequencies in 2025 with traffic volume taken into account and to investigate which traffic and carriageway characteristics are associated with these differences.

## Research Question
**Which highway carriageway sections in South Holland have the highest crash rates in 2025, given how busy they are and which traffic and carriageway characteristics are associated with differences in crash frequency?**


**Subquestions:**
1. How does the ranking of carriageway sections by number of accidents differ from the ranking by risk (accidents per million vehicle-kilometres)?
2. ⁠To what extent does the number of accidents increase with vehicle-kilometres, and which characteristics (carriageway type, share of heavy goods vehicles, night-time traffic) are associated with more accidents than expected?
3. Are the carriageway sections with a highest risk of injury and fatal accidents the same as those with the highest risk of all accidents?

# Data Used

To find road sections that are genuinely unsafe, we need two things: how many accidents happen on each section (the numerator) and how much traffic drives there (the denominator). The remaining data is needed to link these two together and to explain the results. All datasets are open data published by Rijkswaterstaat and are shared in our repository.

| 1 | **BRON – Bestand geRegistreerde Ongevallen Nederland** (accident data, 2025) | Rijkswaterstaat |  **Numerator.** One record per registered accident, with its severity (fatal / injury / property damage only) and its location on the road network (`WVK_ID` or junction `JTE_ID`) | [data.overheid.nl – BRON](https://data.overheid.nl/dataset/9841-verkeersongevallen---bestand-geregistreerde-ongevallen-nederland) |

**The key link between the datasets** is the column `NWB_IDS` in the INWEVA carriageway file. It lists the NWB road sections (`WVK_ID`) that make up each INWEVA carriageway section (`VBN_ID`). The accident file uses the same `WVK_ID`, which allows a direct table join without any spatial matching.

**Files used:**

*BRON*
- **Accidents:** `ongevallen.txt` (all registered accidents in 2025 and their location), with the reference table `aflopen3.txt` to translate severity codes (DOD = fatal, LET = injury, UMS = property damage only)
- **Road network:** `wegvakken.txt` (road sections, road numbers and directions), `hectointervallen.txt` (kilometre positions), `juncties.txt` (junctions), `wegvakgeografie.shp` (map of road sections)

*INWEVA*
- **Carriageway sections:** `inweva_verkeersbanen_2025DEC.csv` (section length, road number, carriageway type and the link to NWB via `NWB_IDS`)
- **Traffic volumes:** `inweva_weekdag_2025DEC.csv` (average daily traffic, share of freight traffic and night-time traffic) and `inweva_weekdag_uren_2025DEC.csv` (traffic per hour, used to check the distribution over day and night)
- **Map:** `INWEVA2025.shp` (geometry of the carriageway sections)
- **Documentation:** product description INWEVA 2025 (PDF), explaining the columns



# Data Pipeline

Our analysis follows five main steps:

1. **Data collection**: we use BRON 2025 for registered road accidents, the Nationaal Wegenbestand (NWB) for the road network, and INWEVA 2025 for traffic volumes and carriageway characteristics.
2. **Data cleaning**: each dataset is cleaned separately. We select the national-highway network in South Holland, remove duplicate or unusable records, check missing values, and make sure the different datasets use consistent formats.
3. **Linking the datasets**: the accident data are connected to the road network and subsequently linked to the corresponding INWEVA carriageway sections. This allows each road section to be combined with its accident information and traffic volume. Accidents located at junctions are treated carefully to avoid assigning one accident to multiple road sections.
4. **Calculating accident risk**: for each carriageway section we calculate the number of accidents, the number of injury and fatal accidents, traffic exposure, road-section length, heavy-goods-vehicle share and night-time traffic share. Traffic exposure is calculated using traffic volume, section length and the number of days in the year. Based on this an accident rate per vehicle-kilometre is calculated.
5. **Analysis and visualisation**: the three subquestions are analysed using rankings, maps, descriptive statistics and regression analysis. The final results show which carriageway sections have relatively high accident risk after traffic exposure is taken into account and which characteristics are associated with these differences.

### Geographical scale
