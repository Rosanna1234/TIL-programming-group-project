## Introduction
To ensure safe roads, it is important to know where most accidents happen. However, some roads have more accidents than others because they carry more traffic. To find the roads that are genuinely more dangerous, the number of accidents has to be related to the amount of traffic. Knowing which road sections are relatively unsafe helps road authorities decide where safety measures are most needed. In this project we focus on the national highways (rijkswegen) in South Holland, managed by Rijkswaterstaat.

## Objective
This project combines Dutch accident registrations (BRON) with traffic volumes (INWEVA) to identify which carriageway sections in South Holland have the highest accident risk once traffic volume is taken into account, and which road and traffic characteristics explain these differences.

## Research Question
**Which sections of national highway in South Holland are unsafe given how busy they are, and what characteristics are associated with this?**


**Subquestions:**
1. How does the ranking of carriageway sections by number of accidents differ from the ranking by risk (accidents per million vehicle-kilometres)?
2. ⁠To what extent does the number of accidents increase with vehicle-kilometres, and which characteristics (carriageway type, share of heavy goods vehicles, night-time traffic) are associated with more accidents than expected?
3. Are the carriageway sections with a highest risk of injury and fatal accidents the same as those with the highest risk of all accidents?

# Data Used

To find road sections that are genuinely unsafe, we need two things: how many accidents happen on each section (the numerator) and how much traffic drives there (the denominator). The remaining data is needed to link these two together and to explain the results. All datasets are open data published by Rijkswaterstaat and are shared in our repository.

| # | Dataset | Publisher | What we use it for | Link |
|---|---|---|---|---|
| 1 | **BRON – Bestand geRegistreerde Ongevallen Nederland** (accident data, 2025) | Rijkswaterstaat | **Numerator.** One record per registered accident: severity (property damage only / injury / fatal), road section ID (`WVK_ID`), hectometre position and province and a file with the parties involved (`partijen`) | [data.overheid.nl – BRON](https://data.overheid.nl/dataset/9841-verkeersongevallen---bestand-geregistreerde-ongevallen-nederland) |
| 2 | **BRON network data** (NWB road sections delivered with BRON 2025) | Rijkswaterstaat | **Selection.** Road manager (`WEGBEHSRT`), road number, carriageway type (`BST_CODE`: main carriageway, on-ramp, off-ramp), kilometre positions (`hectointervallen`), junctions and geometry for mapping(`wegvakgeografie.shp`) | Part of the BRON download above |
| 3 | **INWEVA 2025 – Intensiteiten op Wegvakken** | Rijkswaterstaat | **Denominator** Average daily traffic volume per carriageway section, split by vehicle class (light/medium/heavy) and period (day/evening/night), section length, road number and the list of NWB section IDs (`NWB_IDS`), shapefile with geometry | [data.overheid.nl – INWEVA](https://data.overheid.nl/dataset/84461-intensiteit-wegvakken--inweva-#panel-resources) |

**The key link between the datasets** is the column `NWB_IDS` in the INWEVA carriageway file. It lists the NWB road sections (`WVK_ID`) that make up each INWEVA carriageway section (`VBN_ID`). The accident file uses the same `WVK_ID`, which allows a direct table join without any spatial matching.

**Files used:**

- BRON: `ongevallen.txt`, `partijen.txt`, `wegvakken.txt`, `hectointervallen.txt`, `juncties.txt`, `wegvakgeografie.shp`, reference tables (code lists for severity, carriageway types, etc.)
- INWEVA: `inweva_verkeersbanen_2025DEC.csv`, `inweva_weekdag_2025DEC.csv`, `INWEVA2025.shp`, product description (PDF)



# Data Pipeline

Our analysis follows five main steps:

1. **Data collection**: we download the accident data and road network (BRON) and the traffic volumes (INWEVA) for 2025.
2. **Data cleaning**: each dataset is cleaned separately. Using the road manager, road number and province, we select only accidents on national highways in South Holland, remove records with missing or incorrect values, and convert columns to the correct format.
3. **Linking the datasets**: using `NWB_IDS`, each accident (`WVK_ID`) is linked to the INWEVA carriageway section it belongs to, which gives every accident a known traffic volume.
4. **Calculating accident risk**: for each carriageway section we create one row with the number of accidents, the number of severe accidents, the traffic volume, the length, the carriageway type, the share of heavy goods vehicles and the share of night-time traffic. From this we calculate the vehicle-kilometres (volume × length × 365) and the accident rate: accidents per million vehicle-kilometres.
5. **Analysis and visualisation**: we answer the three subquestions as described below, and present the results using maps, charts and regression analysis.

### Approach per subquestion

**Subquestion 1 (ranking):** we place two rankings side by side, one by number of accidents and one by accident rate, and show on two maps which sections rise and which fall when traffic volume is taken into account.

**Subquestion 2 (frequency):** using a regression model, we test how the number of accidents grows with vehicle-kilometres. We also examine whether carriageway type, share of heavy goods vehicles and night-time traffic are associated with more accidents than expected. Carriageway sections that perform much worse than the model predicts are the real hotspots.

**Subquestion 3 (severity):** we calculate the same accident rate for injury and fatal accidents only, and compare whether this points to the same locations as for all accidents.

Together, the three subquestions answer the main question: which sections of national highway in South Holland are unsafe given how busy they are, and which characteristics are associated with this.
